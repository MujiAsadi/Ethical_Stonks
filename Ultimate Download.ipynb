## Get Column Data from CSV

```python
def get_column_from_csv(file, col_name):
    # Try to get the file and if it doesn't exist issue a warning
    try:
        df = pd.read_csv(file)
    except FileNotFoundError:
        print("File Doesn't Exist")
    else:
        return df[col_name]
```

## Save Stock Data to CSV

```python
# Function that gets a dataframe by providing a ticker and starting date
def save_to_csv_from_yahoo(folder, ticker):
    stock = yf.Ticker(ticker)
    
    try:
        print("Get Data for : ", ticker)
        # Get historical closing price data
        df = stock.history(period="10y")
    
        # Wait 2 seconds
        time.sleep(2)
        
        # Remove the period for saving the file name
        # Save data to a CSV file
        # File to save to 
        the_file = folder + ticker.replace(".", "_") + '.csv'
        print(the_file, " Saved")
        df.to_csv(the_file)
    except Exception as ex:
        print("Couldn't Get Data for :", ticker)
```

## Download All Stocks

```python
def batch_download_stocks(ticker_list, folder_path, batch_size=100):
    os.makedirs(folder_path, exist_ok=True)
    
    # Process tickers in chunks to prevent server overloading
    for i in range(0, len(ticker_list), batch_size):
        chunk = ticker_list[i:i + batch_size]
        print(f"Downloading batch {i // batch_size + 1} ({len(chunk)} tickers)...")
        
        try:
            # Download bulk data in parallel
            data = yf.download(
                tickers=list(chunk),
                period="10y",
                group_by="ticker",
                threads=True,
                progress=False
            )
            
            # Save individual non-empty CSVs
            for ticker in chunk:
                try:
                    # Check if multi-index or single ticker output
                    if len(chunk) == 1:
                        df = data
                    else:
                        df = data[ticker].dropna(how="all") if ticker in data else pd.DataFrame()
                    
                    if not df.empty:
                        file_name = ticker.replace(".", "_") + ".csv"
                        file_path = os.path.join(folder_path, file_name)
                        df.to_csv(file_path)
                    else:
                        print(f"Skipping empty or delisted symbol: {ticker}")
                except Exception as ex:
                    print(f"Error processing {ticker}: {ex}")
                    
        except Exception as batch_ex:
            print(f"Batch failed: {batch_ex}")
            
        # Pause briefly between batches to avoid rate limits
        time.sleep(2)
```

```python
# Example execution
tickers = get_column_from_csv("/Users/muji/Documents/Python4Finance/Wilshire-5000-Stocks-New.csv", "Ticker")
batch_download_stocks(tickers, "Wilshire_Stocks/", batch_size=100)
```
